```python
import os 
from dotenv import load_dotenv, find_dotenv

_ = load_dotenv(find_dotenv())

openai_api_key = os.environ["OPENAI_API_KEY"]

print(f"openai_api_key: {openai_api_key}")
```

```python
# Completion model

from langchain_openai import ChatOpenAI, OpenAI

llmModel = OpenAI()
chatModel = ChatOpenAI(model="gpt-3.5-turbo-0125")
```

```python
# llmModel
```

```python
prompt = "Tell me one fact about Kennedy family"
```

```python
response = llmModel.invoke(prompt)
```

```python
print(response)
```

```python
for chunk in llmModel.stream(prompt):
    print(chunk, end="", flush=True)
```

```python
creative_llm_model = OpenAI(temperature=0.9)
```

```python
formal_llm_model = OpenAI(temperature=0.1)
```

```python
response = creative_llm_model.invoke(prompt)
print(response)
```

```python
response = formal_llm_model.invoke(prompt)
print(response)
```

```python
messages = [
    (
        "system", "You are an historian expert in the Kennedy family."
    ),
    (
        "human", "Tell me one interesting thing about JFK"
    )
]

response = chatModel.invoke(messages)
```
