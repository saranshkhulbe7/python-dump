```python
from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_groq import ChatGroq

openai_chat_model = ChatOpenAI(model="gpt-4.1")
# genai_chat_model = ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
# groq_chat_model = ChatGroq(model="qwen/qwen3-32b")
```

```python
# Streaming
gen = openai_chat_model.stream("Write a 200 words paragraph on artificial intelligence.")
```

```python
# while True:
#     try:
#         item = next(gen)
#         print(item)
#     except StopIteration:
#         break
    
for x in gen:
    print(x.text, end=" ", flush=True)
```

```python
# Batch

responses = openai_chat_model.batch(
    [
        "Why do parrots have colorful feathers?",
        "How do airplanes fly?",
        "What is quantum computing?",
    ]
)

for response in responses:
    print(response)
```

```python
responses = openai_chat_model.batch(
    [
        "Why do parrots have colorful feathers?",
        "How do airplanes fly?",
        "What is quantum computing?",
    ],
    config={
        "max_concurrency": 5
    }
)

for response in responses:
    print(response)
```
