```python
import os
from dotenv import load_dotenv, find_dotenv
_=load_dotenv(find_dotenv())
openai_api_key = os.environ["OPENAI_API_KEY"]
print(f"openai_api_key: {openai_api_key}")
```

```python
from langchain_openai import ChatOpenAI
chat_model = ChatOpenAI(model="gpt-4.1")
```

```python
from langchain.tools import tool
from langchain.messages import ToolMessage

@tool
def get_weather(location: str)-> str:
    """Get weather at a location"""
    return f"Its sunny in {location}"

model_with_tools = chat_model.bind_tools([get_weather])

tool_registry = {"get_weather": get_weather}
```

```python
# Tool Execution Loop

messages = [{"role": "user", "content": "What is the weather in Boston"}]
ai_msg = model_with_tools.invoke(messages)
messages.append(ai_msg)

for tool_call in ai_msg.tool_calls:
    result = tool_registry[tool_call["name"]].invoke(tool_call["args"])
    messages.append(ToolMessage(content=result, tool_call_id=tool_call["id"]))


print(messages)
final_result = model_with_tools.invoke(messages)
print(final_result.text)
```
