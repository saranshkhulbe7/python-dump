```python
import os
from dotenv import load_dotenv, find_dotenv

_ = load_dotenv(find_dotenv())
openai_api_key = os.environ["OPENAI_API_KEY"]
print(f"openai_api_key: {openai_api_key}")
```

```python
from langchain_openai import ChatOpenAI
chat_model = ChatOpenAI(model="gpt-4.1")
chat_model
```

```python
from pydantic import BaseModel, Field

class Actor(BaseModel):
    name: str
    role: str

class Movie(BaseModel):
    title: str=Field(description="The title of the movie.")
    year: int=Field(description="This is the year the movie was released.")
    cast: list[Actor]
    genres: list[str]
    director: str=Field(description="The director of the movie.")
    rating: float=Field(description="The movies' rating out of 10")
```

```python
model_with_structure = chat_model.with_structured_output(Movie)
model_with_structure
```

```python
model_with_structure.invoke("Provide details of the movie 'Inception'.")
```

```python
from typing_extensions import TypedDict, Annotated

class Movie(TypedDict):
    """A movie with details"""
    title: Annotated[str, ..., "The title of the movie"]
    year: Annotated[int, ..., "This is the year the movie was released."]

model_with_typed_dict = chat_model.with_structured_output(Movie)
```

```python
model_with_typed_dict.invoke("Provide details of the movie 'Inception'.")
```
